```python
import random

pos = 0

opciones = [[0, 1, 2], [3, 4, 5], [6, 7, 8], [0, 4, 8], [2, 4, 6], [0, 3, 6], [1, 4, 7], [2, 5, 8]]

victorias = {}

class Tablero:
  def __init__(self):
    self.tablero = [1, 2, 3, 4, 5, 6, 7, 8, 9]

  def mostrar_tablero(self, con_numeros=True):
    if con_numeros == True:
      d = 0
      for x in self.tablero:
         print(f"{x}\\ ", end="")
         d += 1
         if d == 3:
          print()
          print("-"*20)
          print()
          d = 0

  def colocar_ficha(self, posicion, simbolo):
    if self.tablero[posicion - 1] in ["X", "O"]:
      return False
    self.tablero[posicion - 1] = simbolo
    return True

  def verificar_ganador(self):
    w = False
    for x in opciones:
      tabo = []
      for y in x:
        tabo.append(self.tablero[y])
      if tabo[0] == tabo[1]:
        if tabo[1] == tabo[2]:
           w = True
    return w

  def tablero_lleno(self):
    a = ["X", "O"]
    nu = 0
    for x in self.tablero:
      if x in a:
        nu += 1
    if nu == 9:
      return True
    else:
      return False

  def reiniciar(self):
    k = 0
    for x in self.tablero:
      self.tablero[k] = k+1
      k += 1

class Jugador:
  def __init__(self, nombre, simbolo):
    self.nombre = nombre
    self.simbolo = simbolo

n1 = input("Digite el nombre del Jugador 1")
s1 = input("Digite el simbolo del Jugador 1 (X/O)")
while s1 != "X" and s1 != "O":
  s1 = input("Digite el simbolo del Jugador 1 (X/O)")
if s1 == "X":
  s2 = "O"
else:
  s2 = "X"
n2 = input("Digite el nombre del Jugador 2")
p1 = Jugador(n1, s1)
p2 = Jugador(n2, s2)
t = Tablero()

victorias = {p1.nombre: 0, p2.nombre: 0}

while True:

  while True:
    try:
      opc = int(input(f"1. Jugar\n2. Ver historial\n3. Salir\n"))
      break
    except ValueError:
      print("Valor invalido")
      pass

  if opc == 1:
    t.reiniciar()
    emp = random.randint(1, 2)
    if emp == 1:
      while True:
        print(f"{p1.nombre} empieza")
        t.mostrar_tablero()
        j = p1.simbolo
        nomb = p1.nombre
        while True:
          try:
            pos = int(input(f"{nomb}, en que posicion desea jugar"))
          except ValueError:
            print("Valor invalido")
            continue
          if pos < 1 or pos > 9:
            print("Jugada no valida")
            continue
          if t.colocar_ficha(pos, j):
            break
          print("Jugada no valida: esa posicion ya esta ocupada")

        veri = t.verificar_ganador()
        if veri == True:
          print(f"{p1.nombre} gana")
          victorias[p1.nombre] += 1
          break
        table = t.tablero_lleno()
        if table == True:
          print("Empates")
          break

        t.mostrar_tablero()
        j = p2.simbolo
        nomb = p2.nombre
        while True:
          try:
            pos = int(input(f"{nomb}, en que posicion desea jugar"))
          except ValueError:
            print("Valor invalido")
            continue
          if pos < 1 or pos > 9:
            print("Jugada no valida")
            continue
          if t.colocar_ficha(pos, j):
            break
          print("Jugada no valida: esa posicion ya esta ocupada")

        veri = t.verificar_ganador()
        if veri == True:
          print(f"{p2.nombre} gana")
          victorias[p2.nombre] += 1
          break
        table = t.tablero_lleno()
        if table == True:
          print("Empates")
          break

    if emp == 2:
      while True:
        print(f"{p2.nombre} empieza")
        t.mostrar_tablero()
        j = p2.simbolo
        nomb = p2.nombre
        while True:
          try:
            pos = int(input(f"{nomb}, en que posicion desea jugar"))
          except ValueError:
            print("Valor invalido")
            continue
          if pos < 1 or pos > 9:
            print("Jugada no valida")
            continue
          if t.colocar_ficha(pos, j):
            break
          print("Jugada no valida: esa posicion ya esta ocupada")

        veri = t.verificar_ganador()
        if veri == True:
          print(f"{p2.nombre} gana")
          victorias[p2.nombre] += 1
          break
        table = t.tablero_lleno()
        if table == True:
          print("Empates")
          break

        t.mostrar_tablero()
        j = p1.simbolo
        nomb = p1.nombre
        while True:
          try:
            pos = int(input(f"{nomb}, en que posicion desea jugar"))
          except ValueError:
            print("Valor invalido")
            continue
          if pos < 1 or pos > 9:
            print("Jugada no valida")
            continue
          if t.colocar_ficha(pos, j):
            break
          print("Jugada no valida: esa posicion ya esta ocupada")

        veri = t.verificar_ganador()
        if veri == True:
          print(f"{p1.nombre} gana")
          victorias[p1.nombre] += 1
          break
        table = t.tablero_lleno()
        if table == True:
          print("Empates")
          break
  if opc == 2:
    print(f"{p1.nombre}: {victorias[p1.nombre]} victorias\n{p2.nombre}: {victorias[p2.nombre]} victorias")

  if opc == 3:
    break
```
